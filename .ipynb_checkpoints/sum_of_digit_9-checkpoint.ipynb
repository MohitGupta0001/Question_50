```python
def sum_of_digits(num):
    """
    Retuen the sum of digits.
    Args:Take input only nubmeric value.
    Return:Return sum of digits.
    """
    if len(num)<2:
        raise ValueError("Please enter minimum two digits for sum .")
    num=int(num)
    if not isinstance(num,(int,float)):
        raise ValueError("Nubmeric value allowed only.")
    
    num = abs(num)
    total = 0
    while num > 0:
        total += num % 10
        num //= 10
    return total
if __name__ == "__main__":
    try:
        num=input("Enter the digits:")
        sum=sum_of_digits(num)
        print("Sum of digits:",sum)
    except ValueError as e:
        print("Error:",e)
    finally:
        print("Program excuted successfully.")
```
