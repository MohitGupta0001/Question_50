```python
def second_largest_number(numbers: list) -> int:
    """
    Return second largest number from a list.
    Args(list): Take input only list of numbers.
    Return: return second largest number in int.
    """
    if len(numbers) < 2:
        raise ValueError("List must contain at least two elements.")
    largest = numbers[0]
    second_largest = numbers[0]
    for i in numbers:
        if i > largest:
            second_largest = largest
            largest = i
        elif i > second_largest and i != largest:
            second_largest = i
    if largest == second_largest:
        raise ValueError("Second largest number does not exist.")
    return second_largest
if __name__ == "__main__":
    try:
        numbers = [1, 5, 8, 40, 600, 70]
        result = second_largest_number(numbers)
        print("Second largest number:", result)
    except ValueError as e:
        print("Error:", e)
    except Exception as e:
        print("Unexpected Error:", e)
    finally:
        print("Program run successfully.")
```
