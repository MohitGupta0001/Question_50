```python
def remove_duplicate_value(list1:list)->list:
    """
    Return the new list without duplicate value.
    Args(list):Take input only list.
    Return(without_duplicate_list):Return the new list without duplicate value.
    """
    if not isinstance(list1,list):
        raise ValueError("We can take input only list.")
    new_list=[]
    for i in list1:
        if i not in new_list:
            new_list.append(i)
    return new_list

if __name__ == "__main__":
    try:
        list1=[1,2,5,4,5,44,54,545,45,44,5,112,44,441]
        without_duplicate_list=remove_duplicate_value(list1)
        print("New list without duplicate values:",without_duplicate_list)
    except ValueError as e:
        print("Error:",e)
    finally:
        print("Program excuted successfully.")
```
