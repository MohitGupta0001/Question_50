```python
def count_vowels(text:str)->int:
    """
    Return text to vowels counts.
    Args(str):Take input only string.
    Return:return count of vowels in int.
    """
    count=0
    vowels="aeiou"
    if not isinstance(text,str):
        raise TypeError("Please enter only string not anything allow.")
    if text.strip() == "":
        raise TypeError("Not enter empty and space in string.")
    for char in text.lower():
        if char in vowels:
            count=count+1
    return count
if __name__ == "__main__":
    try:
        text=input("Enter the string:")
        vowels_in_text=count_vowels(text)
        print("Vowel count:",vowels_in_text)
    except TypeError as e:
        print("Error:",e)
    except Exception as e:
        print(e)
    finally:
        print("Program run successfullly.")
    
```
